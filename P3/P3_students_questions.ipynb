## 0.2 Data Cleaning

**Do the following:** 
1. Fill missing embarked with the most frequent value.
2. Convert sex and class to categorical type.
3. Create a new column family_size= sibsp + parch + 1.
4. Cap fare at the 99th percentile to reduce outlier impact.

```python
# 1
most_freq_value = df['embarked'].mode()[0]
print(most_freq_value)
df['embarked'] = df['embarked'].fillna(most_freq_value)
df['embarked'].isna().sum()
```

```python
# 2
df['sex'] = df['sex'].astype('category')
df['class'] = df['class'].astype('category')
df.dtypes
```

```python
# 3
df['family_size'] = df['sibsp'] + df['parch'] + 1
df.head()
```

```python
# 4
perc_99 = df['fare'].quantile(0.99)
print(perc_99)
df.loc[df['fare'] > perc_99, 'fare'] = perc_99
df['fare'].max()
```
