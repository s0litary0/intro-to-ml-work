## Sorting and Filtering

### 11. How many products cost more than $10.00?

```python
df.item_price > 10.00
```

```python
df[df.item_price > 10.00]["order_id"].count()
```

### 13. Sort the dataset by item

```python
df.sort_values(by="item_name")
```

### 14. What is the quantity of the most expensive item ordered?

```python
df.sort_values(by="item_price", ascending=False)["quantity"].iloc[0]
```

```python
df.loc[df["item_price"].idxmax()]["quantity"]
```

### 15. How many times was a veggie salad Bowl ordered?


```python
df[df.item_name == "Veggie Salad Bowl"]["quantity"].sum()
```

### 16. How many times did someone order more than one canned soda?

```python
condition = (df.quantity > 1) & (df.item_name == "Canned Soda")
```

```python
df[condition]["order_id"].count()
```

### 17. How many different products are sold?

```python
len(df.item_name.unique())
```

```python
df.item_name.nunique()
```

```python
df["item_name"].duplicated()
```

### 18. What is the total revenue?

```python
total_revenue = (df.item_price * df.quantity).sum()
```

```python
total_revenue
```

### 22. Which item has the highest average price?

```python
df.groupby(by="item_name")["item_price"].mean().idxmax()
```

### 23. How many items include chicken in their item name?

```python
df[df.item_name.str.contains("Chicken")]["order_id"].count()
```

## Questions

### 31. What was the most ordered items in the choice_description column?

```python
def get_choice_list(value: str):
    return value.replace("[", "").replace("]", "").split(", ")
```

```python
df.choice_description[df.choice_description.notna()].apply(get_choice_list).value_counts().index[0]
```
